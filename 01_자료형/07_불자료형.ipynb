```python
print(1==1)
print(1>2)
print(1<2)

bool(1==1)
print(bool(["python"])) # 비어 있지 않은 문자열은 true다
print(bool("")) # 빈 문자열은 false다
print(bool(10)) # 숫자는 0이 아닌 이상 true다
print(bool(0)) # 0은 false다
print(bool([1,2,3])) # 비어 있지 않은 리스트는 true다
print(bool([])) # 빈 리스트는 false다
```

```python
## 논리 연산자
# and (그리고)
# or (또는)

print(True and True) # True
print(True and False) # False
print(False and True) # False
print(False and False) # False

print(True or True) # True
print(True or False) # True
print(False or True) # True
print(False or False) # False

## 비교 연산자
# >(크다,초과)
# <(작다,미만)
# >=(크거나 같다,이상)
# <=(작거나 같다,이하)
# == (같다)
# != (같지 않다, 다르다)
```
