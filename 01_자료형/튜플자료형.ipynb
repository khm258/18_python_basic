```python
## 튜플에 요소값을 지우거나 변경할려고 하면?
t1= (1, 2, "a", "b")
del t1[2] # 튜플은 삭제 불가
```

```python
t1= (1, 2, "a", "b")
t1[2] = "c" # 튜플은 변경 불가
```

```python
## 튜플인덱싱
t1= (1, 2, "a", "b")
print(t1[0])  # 1
print(t1[2])  # "a"
```

```python
## 튜플슬라이싱
t1= (1, 2, "a", "b")
print(t1[1:])  # 1
```

```python
## 튜플 더하기
t1= (1, 2, "a", "b")
t2= (3, 4)
t3= t1 + t2
print(t3)  # (1, 2, "a", "b", 3, 4, "c", "d")
```

```python
## 튜플 곱하기
t2=(3,4)
t4= t2 * 2
print(t4)  # (3, 4, 3, 4)
```

```python
## 튜플 길이 구하기 len()
t1= (1, 2, "a", "b")
len(t1)  # 4
```
