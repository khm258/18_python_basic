```python
## 딕셔너리는 어떻게 만들까?
## {key1 : value1, key2 : value2, ...}
dict= {
    "name": "kim",
    "phone": "010-1234-5678",
    "birth": "09-02"
}

print(type(dict))
```

```python
## 딕셔너리 쌍 추가하기
a= {1: "a"}
a[2] = "b"
print(a)  # {1: "a", 2: "b"}

a["name"] = "홍길동"
print(a)  # {1: "a", 2: "b", "name": "홍길동"}

a[3]= [1,2,3]
print(a)  # {1: "a", 2: "b", "name
```

```python
a= {
    1: "a",
    2: "b",
    "name": "홍길동",
    3: [1,2,3],
    (1,2):(1,2),
    # [1,2]:[1,2] <- 리스트는 key로 사용할 수 없다. (TypeError: unhashable type: 'list')

}
print(a)  # {1: "a", 2: "b", "name": "홍길동", 3: [1,2,3], (1,2):(1,2), [1,2]:[1,2]}
```

```python
## 딕셔너리에서 key를 사용하여 value 얻기
학점 ={
    "홍길동": 85,
    "김철수": 90,
}
print(학점["김철수"])
print(학점["홍길동"])

학점["홍길동"] = 95
학점["강호동"] = 75
print(학점)
```

```python
## 딕셔너리 관련 함수
## key 리스트 만들기 keys() 함수
dict= {
    "name": "kim",
    "phone": "010-1234-5678",
    "birth": "09-02"
}
print(dict.keys())  # dict_keys(['name', 'phone', 'birth'])
print(list(dict.keys()))  # ['name', 'phone', 'birth']
```

```python
## value 리스트 만들기 values() 함수
dict= {
    "name": "kim",
    "phone": "010-1234-5678",
    "birth": "09-02"
}
print(dict.values())  # dict_values(['kim', '010-1234-5678', '09-02'])
print(list(dict.values()))  # ['kim', '010-1234-5678', '09-02']
```

```python
## key, value 쌍 모두 지우기 clear() 함수
dict= {
    "name": "kim",
    "phone": "010-1234-5678",
    "birth": "09-02"
}
dict.clear()
print(dict)  # {}
```

```python
## key로 value 얻기 get() 함수
dict= {
    "name": "kim",
    "phone": "010-1234-5678",
    "birth": "09-02"
}
print(dict.get("name"))  # kim
```
