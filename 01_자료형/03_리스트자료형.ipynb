```python
# 리스트는 어떻게 만들고 사용할까?
# 리스트명 = [요소1, 요소2, 요소3, ...]

a=[]
b= [1, 2, 3]
c= ["Life", "is", "too", "short"]
d= [1, 2, ["Life", "is"]]
e= [1, 2, "Life", "is"]
f= list() # 빈 리스트 만들기

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
## 리스트에 인덱싱
a= [1, 2, 3]
print(type(a))
print(a[0])  # 첫 번째 요소 출력
print(a[1])  # 두 번째 요소 출력
print(a[2])  # 세 번째 요소 출력
```

```python
a= [1, 2, 3, ["a", "b", "c"]]
len(a)  # 리스트의 길이 출력
print(a[3])  # 네 번째 요소 출력
print(a[3][0])  # 네 번째 요소의 첫 번째 하위 요소 출력
print(a[3][1])  # 네 번째 요소의 두 번째 하위 요소 출력
print(a[3][2])  # 네 번째 요소의 세 번째 하위 요소 출력
```

```python
## 리스트의 슬라이싱
a= [1, 2, 3, 4, 5]
print(a[0:2])  # 첫 번째 요소부터 두 번째 요소까지 출력
print(a[2:])  # 세 번째 요소부터 마지막 요소까지 출력
print(a[:2])  # 첫 번째 요소부터 두 번째 요소까지 출력
print(a[:])  # 모든 요소 출력
```

```python
a= [1, 2, 3, ["a", "b", "c"]]
print(a[3][1:3]) # 세 번째 요소의 두 번째 하위 요소부터 세 번째 하위 요소까지 출력
```

```python
## 리스트의 연산
a= [1, 2, 3]
b= [4, 5, 6]
print(a+b)  # 리스트 a와 b를 연결
print(a*3)  # 리스트 a를 세 번 반복
```

```python
## 리스트 길이 구하기 len
a= [1, 2, 3]
print(len(a))  # 리스트의 길이 출력
```

```python
## 리스트의 수정과 삭제
a= [1, 2, 3]
a[2] = 4  # 세 번째 요소 수정
print(a)
del a[1]  # 두 번째 요소 삭제
print(a)
```

```python
## 리스트 관련 함수

## 리스트의 요소 추가하기 append
a= [1, 2, 3]
a.append(4)  # 리스트 a의 마지막에 4를 추가
print(a)
a.append(["a", "b"])  # 리스트 a의 마지막에 ["a", "b"]을 추가
print(a)
```

```python
## 리스트정렬 sort
a= [1, 4, 3, 2]
a.sort()  # 리스트 a를 오름차순으로 정렬
print(a)  # [1, 2, 3, 4]

a= ["a", "c", "b"]
a.sort()  # 리스트 a를 오름차순으로 정렬
print(a)  # ['a', 'b', 'c']

## 내림차순
a.sort(reverse=True)  # 리스트 a를 내림차순으로 정렬
print(a)  # ['c', 'b', 'a']
```

```python
## 리스트 뒤집기 = reverse
a= ["a", "c", "b"]
a.reverse()  # 리스트 a를 뒤집기
print(a)  # ['b', 'c', 'a']
```

```python
## 인덱스 반환 = index
a= [1, 2, 3]
print(a.index(2))  # 1 # 리스트 a에서 2의 인덱스 반환
```

```python
## 리스트 삽입 insert
a= [1, 2, 3]
a.insert(1, 5)  # 리스트 a의 인덱스 1에 5를 삽입
print(a)  # [1, 5, 2, 3]
```

```python
## 리스트 요소 제거 remove
a= [1, 2, 3, 2]
a.remove(2)  # 리스트 a에서 2를 제거
print(a)  # [1, 3, 2]
```

```python
## 리스트 요소 꺼내기 pop
a= [1, 2, 3]
element = a.pop()  # 리스트 a의 마지막 요소를 꺼내기
print(element)  # 3
print(a)  # [1, 2]
```

```python
## 리스트에 포함된 x의 개수 세기 count
a= [1, 2, 3, 1]
print(a.count(1))  # 2 # 리스트 a에서 1의 개수 반환
```

```python
## 리스트 확장 extend
a= [1, 2, 3]
b= [4, 5]
a.extend(b)  # 리스트 a에 리스트 b를 확장
print(a)  # [1, 2, 3, 4, 5]
```
