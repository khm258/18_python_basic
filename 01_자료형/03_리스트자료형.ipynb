```python
# 리스트는 어떻게 만들고 사용할까?
# 리스트명 = [요소1, 요소2, 요소3, ...]

a=[]
b= [1, 2, 3]
c= ["Life", "is", "too", "short"]
d= [1, 2, ["Life", "is"]]
e= [1, 2, "Life", "is"]
f= list() # 빈 리스트 만들기

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
## 리스트에 인덱싱
a= [1, 2, 3]
print(type(a))
print(a[0])  # 첫 번째 요소 출력
print(a[1])  # 두 번째 요소 출력
print(a[2])  # 세 번째 요소 출력
```

```python
a= [1, 2, 3, ["a", "b", "c"]]
len(a)  # 리스트의 길이 출력
print(a[3])  # 네 번째 요소 출력
print(a[3][0])  # 네 번째 요소의 첫 번째 하위 요소 출력
print(a[3][1])  # 네 번째 요소의 두 번째 하위 요소 출력
print(a[3][2])  # 네 번째 요소의 세 번째 하위 요소 출력
```

```python
## 리스트의 슬라이싱
a= [1, 2, 3, 4, 5]
print(a[0:2])  # 첫 번째 요소부터 두 번째 요소까지 출력
print(a[2:])  # 세 번째 요소부터 마지막 요소까지 출력
print(a[:2])  # 첫 번째 요소부터 두 번째 요소까지 출력
print(a[:])  # 모든 요소 출력
```

```python
a= [1, 2, 3, ["a", "b", "c"]]
print(a[3][1:3]) # 세 번째 요소의 두 번째 하위 요소부터 세 번째 하위 요소까지 출력
```

```python
## 리스트의 연산
a= [1, 2, 3]
b= [4, 5, 6]
print(a+b)  # 리스트 a와 b를 연결
print(a*3)  # 리스트 a를 세 번 반복
```

```python
## 리스트 길이 구하기 len
a= [1, 2, 3]
print(len(a))  # 리스트의 길이 출력
```

```python
## 리스트의 수정과 삭제
a= [1, 2, 3]
a[2] = 4  # 세 번째 요소 수정
print(a)
del a[1]  # 두 번째 요소 삭제
print(a)
```

```python
## 리스트 관련 함수

## 리스트의 요소 추가하기 append
a= [1, 2, 3]
a.append(4)  # 리스트 a의 마지막에 4를 추가
print(a)
a.append(["a", "b"])  # 리스트 a의 마지막에 ["a", "b"]을 추가
print(a)
```
