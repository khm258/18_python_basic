```python
# 집합자료형은 어떻게 만들까?
## 집합자료형의 특징
## 1. 중복을 허용하지 않는다.
## 2. 순서가 없다.

s1= set([1,2,3])
s2= set("Hello")
s3= {1,2,3}
s4={}  
s5= set()  # 빈 집합을 만들 때는 set()을 사용해야 한다. {}는 딕셔너리 자료형으로 인식된다.

print(s1, type(s1))
print(s2, type(s2))
print(s3, type(s3))
print(s4, type(s4))
print(s5, type(s5))
```

```python
## 교집합
s1= {1,2,3,4,5,6}
s2= {4,5,6,7,8,9}
print(s1 & s2)  # 교집합
print(s1.intersection(s2))  # 교집합
```

```python
## 합집합
s1= {1,2,3,4,5,6}
s2= {4,5,6,7,8,9}
print(s1 | s2)  # 합집합
print(s1.union(s2))  # 합집합
```

```python
## 차집합
s1= {1,2,3,4,5,6}
s2= {4,5,6,7,8,9}
print(s1 - s2)  # 차집합
```

```python
## 집합자료형 관련 함수
## 값 1개 추가하기

s1= {1,2,3}
s1.add(4)
```

```python
## 값 여러개 추가하기
s1= {1,2,3}
s1.update([4,5,6])
print(s1)
```

```python
## 특정값 제거하기
s1= {1,2,3}
s1.remove(2)  # 특정값 제거하기
print(s1)
```
