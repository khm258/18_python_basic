```python
## 문자열 포메팅
print("i eat %d apples" %3)

# 숫자는 d
```

```python
"i eat %s apples" %"five"
#문자는 s로 한다
print("i eat %d apples" %3)
print("i eat %s apples" %"d")
```

```python
## format 함수를 사용한 포메팅
"i eat {0} apples".format(3)
"i eat {0} apples".format("five")

## 2개 이상의 값을 넣을때
number=10
day="three"
"i ate {0} apples. so I was sick for {1} days.".format(number, day)
```

```python
## f문자열 포메팅
name="홍길동"
age=30

print(type(name))
print(type(age))

print(f"나의 이름은 {name}이고 나이는 {age}입니다.")
```
