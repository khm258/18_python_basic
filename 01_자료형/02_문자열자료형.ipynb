```python
## 문자열 포메팅
print("i eat %d apples" %3)

# 숫자는 d
```

```python
"i eat %s apples" %"five"
#문자는 s로 한다
print("i eat %d apples" %3)
print("i eat %s apples" %"d")
```

```python
## format 함수를 사용한 포메팅
"i eat {0} apples".format(3)
"i eat {0} apples".format("five")

## 2개 이상의 값을 넣을때
number=10
day="three"
"i ate {0} apples. so I was sick for {1} days.".format(number, day)
```

```python
## f문자열 포메팅
name="홍길동"
age=30

print(type(name))
print(type(age))

print(f"나의 이름은 {name}이고 나이는 {age}입니다.")
```

```python
## 문자열 관련 함수들
a= "hobby"
print(type(a))

## 문자 개수 세기 count
print(a.count("b"))
"hobby".count("b")
```

```python
## 위치 알려주기 find
a= "Python is the best choice"
print(a.find("b"))
# 없는값은 -1이라 나온다
print(a.find("k"))
```

```python
## 위치 알려주기 index
a= "Life is too short"
print(a.index("t"))
# find와 다르게 에러가 나온다
print(a.index("k"))
```

```python
## 문자열 삽입 join
",".join(["a", "b", "c", "d"])
```

```python
## 소문자를 대문자로 바꾸기 upper
a= "hi"
print(a.upper())
```

```python
## 대문자 소문자로 바꾸기 lower
a= "HI"
print(a.lower())
```

```python
## 양쪽 공백 지우기 strip
a= "   hi   "
print(a.strip())
```

```python
## 문자열 바꾸기 replace
a= "Life is too short"
print(a.replace("Life", "Leg"))
```

```python
## 문자열 나누기 split
a= "Life is too short"
print(a.split())
```
